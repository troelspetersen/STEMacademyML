```python
# Create a copy to avoid modifying original
df_smeared = df.copy()

# Set random seed for reproducibility
np.random.seed(42)

# Get 10% of samples randomly
sample_size = int(len(df) * 0.1)

# Generate different random indices for each column
random_indices_1 = np.random.choice(df.index, size=sample_size, replace=False)
random_indices_2 = np.random.choice(df.index, size=sample_size, replace=False)
random_indices_3 = np.random.choice(df.index, size=sample_size, replace=False)
random_indices_4 = np.random.choice(df.index, size=sample_size, replace=False)
random_indices_5 = np.random.choice(df.index, size=sample_size, replace=False)
random_indices_6 = np.random.choice(df.index, size=sample_size, replace=False)
random_indices_7 = np.random.choice(df.index, size=sample_size, replace=False)

# Get all columns except 'Age'
columns_to_modify = [col for col in df.columns if col != 'Age' and col != 'Hours of Training']
print(f"Columns to modify: {columns_to_modify}")

# Apply noise to each column (assuming you have 7 columns to modify)
noise_configs = [
    (random_indices_1, 'Resting Heart Rate', 20),  # scale=20 for heart rate
    (random_indices_2, columns_to_modify[1], 30),  # scale=10 for second column
    (random_indices_3, columns_to_modify[2], 30),  # scale=15 for third column
    (random_indices_4, columns_to_modify[3], 25),   # scale=5 for fourth column
    (random_indices_5, columns_to_modify[4], 20),  # scale=12 for fifth column
    (random_indices_6, columns_to_modify[5], 20),   # scale=8 for sixth column
]

# Apply noise to each column
for indices, column, noise_scale in noise_configs:
    if column in df.columns:
        # Create Gaussian noise
        gaussian_noise = np.random.normal(loc=0, scale=noise_scale, size=sample_size)
        
        # Add noise to the selected samples
        df_smeared.loc[indices, column] += gaussian_noise
        
        # Ensure no negative values (adjust this based on your data constraints)
        df_smeared[column] = np.maximum(df_smeared[column], 0)
        
        print(f"Added noise to {column}: {len(indices)} samples modified")

print(f"\nTotal columns modified: {len([col for col in columns_to_modify if col in df.columns])}")
```

```python
# Compare original vs smeared distribution
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(df['Resting Heart Rate'], bins=50, alpha=0.7, label='Original')
plt.title('Original Distribution')
plt.xlabel('Resting Heart Rate')
plt.ylabel('Frequency')
plt.xlim(0,160)

plt.subplot(1, 2, 2)
plt.hist(df_smeared['Resting Heart Rate'], bins=50, alpha=0.7, color='orange', label='Smeared')
plt.title('Smeared Distribution (10% samples + noise)')
plt.xlabel('Resting Heart Rate')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
# Create exponentially decaying distribution for training hours
# Most people train very little, fewer people train a lot

# Set random seed for reproducibility
np.random.seed(42)

# Generate exponential distribution
# Scale controls the decay rate - smaller scale = faster decay
scale = 3.0  # Adjust this to control how quickly it decays
size = len(df)

# Generate exponential distribution (starts high, decays quickly)
training_hours = np.random.exponential(scale=scale, size=size)

# Transform to realistic training hours range (0-20 hours per week)
# Most people will train 0-5 hours, very few will train 15-20 hours
training_hours = np.clip(training_hours, 0, 20)

# Round to realistic values (half-hour increments)
training_hours = np.round(training_hours * 2) / 2

# Add to dataframe
df['Training Hours Per Week'] = training_hours

# Plot the distribution
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(training_hours, bins=40, alpha=0.7, color='skyblue', edgecolor='black')
plt.title('Training Hours Per Week Distribution')
plt.xlabel('Hours per Week')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.3)

plt.subplot(1, 2, 2)
plt.hist(training_hours, bins=40, alpha=0.7, color='lightgreen', cumulative=True, density=True)
plt.title('Cumulative Distribution')
plt.xlabel('Hours per Week')
plt.ylabel('Cumulative Probability')
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

# Print some statistics
print(f"Mean training hours: {training_hours.mean():.2f}")
print(f"Median training hours: {np.median(training_hours):.2f}")
print(f"% training 0-2 hours: {(training_hours <= 2).mean()*100:.1f}%")
print(f"% training 5+ hours: {(training_hours >= 5).mean()*100:.1f}%")
print(f"% training 10+ hours: {(training_hours >= 10).mean()*100:.1f}%")
```
