```python
import pandas as pd
import numpy as np
import os
```

Indsæt navet på et gyldgit ML datasæt og filen returnerer 2 opdelte datasæt, med (som udgangspunkt) 80% og 20% af dataen. Hvis du gerne vil bruge dette til konkurrence så husk at sørge for at fjerne target kolonnen fra det lille datasæt inden du udleverer det.

```python

def split_dataset(file_name, train_ratio=0.8, random_state=42):
    """
    Split dataset into training and test sets
    
    Parameters:
    -----------
    file_name : str
        Name or path of the CSV file
    train_ratio : float
        Ratio of data for training (default 0.8 = 80%)
    random_state : int
        Random seed for reproducibility
    
    Returns:
    --------
    tuple : (train_file_path, test_file_path)
    """
    print("="*70)
    print(f"SPLITTING DATASET: {file_name}")
    print("="*70)
    
    try:
        # Load the dataset
        df = pd.read_csv(file_name)
        total_rows = len(df)
        print(f"✅ Dataset loaded: {total_rows} rows, {len(df.columns)} columns\n")
        
        # Shuffle the dataset
        df_shuffled = df.sample(frac=1, random_state=random_state).reset_index(drop=True)
        
        # Calculate split point
        split_point = int(total_rows * train_ratio)
        
        # Split the data
        df_train = df_shuffled[:split_point]
        df_test = df_shuffled[split_point:]
        
        print("-"*70)
        print("📊 SPLIT INFORMATION")
        print("-"*70)
        print(f"Total rows: {total_rows}")
        print(f"Training set: {len(df_train)} rows ({(len(df_train)/total_rows)*100:.1f}%)")
        print(f"Test set: {len(df_test)} rows ({(len(df_test)/total_rows)*100:.1f}%)")
        print()
        
        # Create output filenames
        file_dir = os.path.dirname(file_name)
        file_base = os.path.basename(file_name)
        file_name_without_ext = os.path.splitext(file_base)[0]
        
        train_filename = os.path.join(file_dir, f"{file_name_without_ext}_train.csv")
        test_filename = os.path.join(file_dir, f"{file_name_without_ext}_test.csv")
        
        # Save the splits
        print("-"*70)
        print("💾 SAVING SPLIT DATASETS")
        print("-"*70)
        
        df_train.to_csv(train_filename, index=False)
        print(f"✅ Training set saved as:")
        print(f"   {train_filename}")
        print()
        
        df_test.to_csv(test_filename, index=False)
        print(f"✅ Test set saved as:")
        print(f"   {test_filename}")
        print()
        
        print("="*70)
        print("✅ DATASET SPLIT COMPLETE!")
        print("="*70)
        
        return train_filename, test_filename
        
    except FileNotFoundError:
        print(f"❌ ERROR: File '{file_name}' not found!")
        print("Please check the file name and path.")
        return None, None
    except Exception as e:
        print(f"❌ ERROR: {str(e)}")
        import traceback
        traceback.print_exc()
        return None, None
```

```python

# Usage:
# Specify your dataset file
file_name = "Diabetes.csv"  # Ændr det her til at være din fil

# Split the dataset (80% train, 20% test)
train_file, test_file = split_dataset(file_name, train_ratio=0.8, random_state=42)  #Den magiske linje

# The function will create two new files:
# - your_dataset_train.csv (80% of data)
# - your_dataset_test.csv (20% of data)
```
